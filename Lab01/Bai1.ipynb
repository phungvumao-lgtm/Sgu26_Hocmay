## Bài 1. Tạo Tensor cơ bản
Tạo một Tensor số nguyên kích thước 2 × 3 và hiển thị kết quả.

```python
import torch
a = torch.tensor(
    [[1,2,3],
     [4,5,6]],
     dtype = torch.int32
)
print(a)
print("kich thuoc",a.shape)
print("so chieu",a.ndim)
print("kieu du lieu",a.dtype)
```

## Bài 2. Tạo các Tensor thường dùng
Thực hành tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1 và ma trận đơn vị.

Nhận xét rand() và randn()

rand()
tạo tensor gồm các số ngẫu nhiên từ 0 đến 1(phân phối đều)

randn() 
tạo tensor gồm các số ngẫu nhiên theo phân phối chuẩn, có thể xuất hiện số âm

```python
import torch

a = torch.rand(3,4)
b = torch.randn(3,4)
c = torch.zeros(2,4)
d = torch.ones(2,4)
e = torch.eye(4)
f = torch.full((5,5),10)

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

## Bài 3. Kiểm tra thông tin Tensor
Sử dụng shape, size(), ndim và dtype để kiểm tra các thuộc tính của Tensor.

torch.Size([2, 3, 4]) cho biết Tensor có 3 chiều, với kích thước lần lượt là 2, 3 và 4. Có thể hiểu Tensor gồm 2 ma trận, mỗi ma trận có 3 hàng và 4 cột, tổng cộng có 24 phần tử.

```python
import torch
a = torch.zeros(2,3,4)
print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

## Bài 4. Slicing và Indexing
Truy xuất một phần dữ liệu từ Tensor nhiều chiều.

a[1] -> (4,5)

a[1:, 2:4] -> (2,2,5)

```python
import torch
a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])

b = torch.randn(3,4,5)
print(b[1].shape)
print(b[1:,2:4].shape)
print("Hai phan tu cuoi cung tren chieu thu ba")
print(b[:, :, -2:])
```
