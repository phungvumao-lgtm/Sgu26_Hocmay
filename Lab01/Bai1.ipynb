• Cho biết shape của a[1] và a[1:, 2:4].

• Thử lấy hai phần tử cuối cùng trên chiều thứ ba.

a[1] -> (4,5)

## Bài 5. Reshape và Flatten
Thay đổi hình dạng Tensor mà không làm thay đổi số phần tử.

Yêu cầu:

• Kiểm tra số phần tử trước và sau khi reshape.

• Thử reshape Tensor thành kích thước 2 × 6.

```python
import torch
a = torch.randn(3, 4)
print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
print("\nso phan tu truoc rasphape")
print(a.numel())
print("\nso phan tu sau rasphape")
print(a.reshape(3 ,2, 2).numel())
print("\nresphape to 2 x 6")
print(a.reshape(2,6))
```

## Bài 6. Phép toán và nhân ma trận
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.

Yêu cầu:

• Phân biệt phép toán a * b với a @ b.T.

• Cho biết shape của kết quả a @ b.T. 

a * b : nhân từng phần tử tương ứng

## Bài 7. Các hàm thống kê trên Tensor
Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

Yêu cầu:

• Thay dim=0 bằng dim=1 và nhận xét sự khác nhau.

dim=0 đi theo chiều dọc tính theo cột
dim=1 đi theo chiều ngang tính theo hàng

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=1))
print("\nStandard deviation:")
print(torch.std(a, dim=1))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
```

## Bài 8. Autograd cơ bản
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

Yêu cầu:

• Tính thủ công đạo hàm của y = x2 tại x = 3.6 và so sánh với x.grad.

• Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

tính thủ công = 7.2, tính x.grad = 7.2 => PyTorch autograd tính gradient chính xác

thay đổi hàm y thủ công bằng 23.6 , tính x.grad = 23.6 => PyTorch autograd tính gradient chính xác



```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = 3*x**2 + 2*x + 1
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

## Bài 9. Gradient Descent tìm hệ số đa thức
Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

Yêu cầu:

• So sánh ba hệ số học được với [1, 2, 3].

• Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().

đúng với ba hệ số của đa thức ban đầu y = x² + 2x + 3.

Trong quá trình Gradient Descent, optimizer.zero_grad() dùng để xóa gradient của lần lặp trước, mse.backward() tính gradient tự động bằng Autograd, còn optimizer.step() cập nhật các hệ số w dựa trên gradient

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

## Bài 10. Giải hệ phương trình bằng Autograd
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

Yêu cầu:

• Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.

• Nhận xét vì sao bài toán có thể có nhiều nghiệm.

Sử dụng Autograd và NAdam, ta tìm được nghiệm thỏa mãn gần đúng cả 4 phương trình. Hệ có vô số nghiệm vì chỉ có 3 phương trình độc lập cho 4 ẩn.

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```
